## A.4. Create a new column
Tạo cột `bmi_group`: Normal (<25), Overweight (25-30), Obese (>=30).

```python
# TODO
def bmi_group(bmi):
    if bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

df["bmi_group"] = df["bmi"].apply(bmi_group)

df.head()
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Người hút thuốc trả chi phí cao hơn bao nhiêu lần so với người không hút, và có đồng đều giữa các vùng không?

```python
# TODO
avg_smoker = df.groupby("smoker")["charges"].mean()

print(avg_smoker)

ratio = avg_smoker["yes"] / avg_smoker["no"]

print("Người hút thuốc trả cao gấp:", ratio, "lần")
```

## Câu hỏi 2: BMI có tương quan với chi phí mạnh hơn ở nhóm hút thuốc hay không hút thuốc?

```python
# TODO
smoker_corr = df[df["smoker"] == "yes"]["bmi"].corr(
    df[df["smoker"] == "yes"]["charges"]
)

non_smoker_corr = df[df["smoker"] == "no"]["bmi"].corr(
    df[df["smoker"] == "no"]["charges"]
)

print("Tương quan nhóm hút thuốc:", smoker_corr)
print("Tương quan nhóm không hút:", non_smoker_corr)

if abs(smoker_corr) > abs(non_smoker_corr):
    print("BMI tương quan mạnh hơn ở nhóm hút thuốc.")
else:
    print("BMI tương quan mạnh hơn ở nhóm không hút thuốc.")
```

## Câu hỏi 3: Vùng nào có chi phí bảo hiểm trung bình cao nhất?

```python
# TODO
result = df.groupby("region")["charges"].mean()

print(result)

print("\nVùng có chi phí cao nhất:")
print(result.idxmax())
```

## Câu hỏi 5: Tuổi có tương quan với chi phí bảo hiểm không?

```python
# TODO
corr = df["age"].corr(df["charges"])

print("Tương quan giữa tuổi và chi phí:", corr)

if corr > 0:
    print("Tuổi có tương quan dương với chi phí bảo hiểm.")
elif corr < 0:
    print("Tuổi có tương quan âm với chi phí bảo hiểm.")
else:
    print("Tuổi không có tương quan với chi phí bảo hiểm.")
```
